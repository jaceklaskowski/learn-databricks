# Genie Agents

(Genie Agents were formerly known as Genie Spaces.)

* [Genie Agents concepts](https://docs.databricks.com/aws/en/genie-agents/concepts)
* [Create and manage a Genie Agent](https://docs.databricks.com/aws/en/genie-agents/set-up)


**Genie Agents** use the natural-language chat interface that business users use to ask questions about their data in Databricks.

Allows business teams to interact with their data using natural language.

Genie Agents convert natural language questions to equivalent SQL queries.

Domain experts configure Genie Agents with datasets (in Unity Catalog), example SQL queries, and text guidelines (instructions), and trusted assets to help Genie translate business questions into analytical queries.

After setup, business users can ask questions and generate visualizations to understand operational data.

In most cases, Genie generates a SQL query that runs on the agent's SQL warehouse.


Genie Agents use data registered in Unity Catalog:
* managed, external and foreign tables.
* views, metric views, and materialized views.
* unstructured data in Unity Catalog volumes (PDFs, Word documents, or other file-based content).

Genie Agents uses the metadata attached to Unity Catalog objects, as well as an author-curated agent-level knowledge store, to generate responses.

The agent can only query tables that you add to the Genie Agent.

**tl;dr** Well-annotated datasets, paired with specific instructions that you provide, are key to creating a positive experience for end users.

## Work Modes

Genie Agents can work in two modes:
1. [Agent mode](https://docs.databricks.com/aws/en/genie-agents/concepts#agent-mode) (formerly known as Research Agent).
1. **Chat mode** that works with structured data only.


## Agent Mode API

You can run Agent mode programmatically through the Agent mode APIs.

See [Agent mode APIs in Genie Agents](https://docs.databricks.com/aws/en/genie-agents/api).
